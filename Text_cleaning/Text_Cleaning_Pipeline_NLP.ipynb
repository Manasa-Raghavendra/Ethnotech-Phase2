```python
import re
import nltk
import contractions
import pandas as pd

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

# Download required NLTK resources
nltk.download('stopwords')
nltk.download('punkt')
nltk.download('punkt_tab')

print("Libraries imported successfully!")
```

```python
# Get English stopwords
stop_words = set(stopwords.words('english'))

def clean_text(text):
    # 1. Convert text to lowercase
    text = text.lower()

    # 2. Expand contractions
    text = contractions.fix(text)

    # 3. Remove hashtags and mentions
    text = re.sub(r'[@#]\w+', '', text)

    # 4. Remove special characters and punctuation
    text = re.sub(r'[^a-zA-Z\s]', ' ', text)

    # 5. Remove extra spaces
    text = re.sub(r'\s+', ' ', text).strip()

    # 6. Tokenize the text
    words = word_tokenize(text)

    # 7. Remove stopwords
    words = [word for word in words if word not in stop_words]

    # 8. Join words back into a sentence
    cleaned_text = ' '.join(words)

    return cleaned_text
```

```python
cleaned_review = clean_text(review)

print("Cleaned Review:")
print(cleaned_review)
```

```python
# Task 2: Social Media Text Cleaning

social_comments = [
    "OMG!!! I can't believe this 😍 @user #Amazing",
    "This movie wasn't good... I don't recommend it!!!",
    "I'm LOVING this product!!! It's sooo good 😍"
]

print("Number of comments:", len(social_comments))

for i, comment in enumerate(social_comments, start=1):
    print(f"\nComment {i}:")
    print(comment)
```

```python
# Clean all social media comments

cleaned_comments = []

for comment in social_comments:
    cleaned = clean_text(comment)
    cleaned_comments.append(cleaned)

for i, cleaned in enumerate(cleaned_comments, start=1):
    print(f"\nCleaned Comment {i}:")
    print(cleaned)
```

```python
# Compare original and cleaned comments

task2_comparison = pd.DataFrame({
    "Original Comment": social_comments,
    "Cleaned Comment": cleaned_comments
})

task2_comparison
```

```python
# Calculate word counts before and after cleaning

words_before_task2 = []
words_after_task2 = []

for original, cleaned in zip(social_comments, cleaned_comments):
    before = len(word_tokenize(original))
    after = len(word_tokenize(cleaned))

    words_before_task2.append(before)
    words_after_task2.append(after)

task2_comparison["Words Before"] = words_before_task2
task2_comparison["Words After"] = words_after_task2

task2_comparison
```

```python
# Overall word reduction

total_before_task2 = sum(words_before_task2)
total_after_task2 = sum(words_after_task2)

total_removed_task2 = total_before_task2 - total_after_task2

reduction_percentage_task2 = (
    total_removed_task2 / total_before_task2
) * 100

print("Total words before cleaning :", total_before_task2)
print("Total words after cleaning  :", total_after_task2)
print("Total words removed         :", total_removed_task2)
print(
    "Reduction percentage        :",
    round(reduction_percentage_task2, 2),
    "%"
)
```

```python
# Count words before preprocessing

df_news["words_before"] = df_news["article"].apply(
    lambda x: len(word_tokenize(x))
)

df_news[["article", "words_before"]].head()
```

```python
# Find the most frequent stopwords in original articles

all_original_words = []

for article in df_news["article"]:
    words = word_tokenize(article.lower())
    all_original_words.extend(words)

# Keep only stopwords

stopword_counts = Counter(
    word for word in all_original_words
    if word in stop_words
)

top_stopwords = stopword_counts.most_common(10)

print("Most frequent stopwords:\n")

for rank, (word, count) in enumerate(top_stopwords, start=1):
    print(f"{rank}. {word} - {count}")
```

```python
# Stopword frequency table

top_stopwords_df = pd.DataFrame(
    top_stopwords,
    columns=["Stopword", "Frequency"]
)

top_stopwords_df.index = range(1, len(top_stopwords_df) + 1)
top_stopwords_df.index.name = "Rank"

top_stopwords_df
```
