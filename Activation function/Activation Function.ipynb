```python
def relu(x):
    if x < 0:
        return 0
    else:
        return x

print("ReLU Activation Function")

for i in range(-10, 11):
    print(f"x = {i:2d} -> {relu(i)}")
```

```python
def leaky_relu(x):
    if x < 0:
        return 0.01 * x
    else:
        return x

print("Leaky ReLU Activation Function")

for i in range(-10, 11):
    print(f"x = {i:2d} -> {leaky_relu(i):.2f}")
```

```python
import math

def elu(x):
    if x >= 0:
        return x
    else:
        return math.exp(x) - 1

print("ELU Activation Function")

for i in range(-10,11):
    print(f"x = {i:2d} -> {elu(i):.6f}")
```

```python
import math

def softplus(x):
    return math.log(1 + math.exp(x))

print("Softplus Activation Function")

for i in range(-10,11):
    print(f"x = {i:2d} -> {softplus(i):.6f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Create x values
x = np.linspace(-10, 10, 400)

# Activation Functions
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def tanh(x):
    return np.tanh(x)

def relu(x):
    return np.maximum(0, x)

def leaky_relu(x):
    return np.where(x > 0, x, 0.01 * x)

def elu(x):
    return np.where(x >= 0, x, np.exp(x) - 1)

def swish(x):
    return x / (1 + np.exp(-x))

# Plot all functions
plt.plot(x, sigmoid(x), label="Sigmoid")
plt.plot(x, tanh(x), label="Tanh")
plt.plot(x, relu(x), label="ReLU")
plt.plot(x, leaky_relu(x), label="Leaky ReLU")
plt.plot(x, elu(x), label="ELU")
plt.plot(x, swish(x), label="Swish")

plt.title("Activation Functions")
plt.xlabel("Input")
plt.ylabel("Output")
plt.grid(True)
plt.legend()

plt.show()
```

# Task 4: Comparison of Activation Functions

| Activation Function | Formula | Output Range | Differentiable | Advantages | Disadvantages | Typical Use |
|---------------------|---------|--------------|----------------|------------|---------------|-------------|
| Binary Step | f(x)=0 if x<0 else 1 | 0 or 1 | No | Very simple and fast | Cannot use backpropagation | Perceptron |
| Linear (Identity) | f(x)=x | (-∞,∞) | Yes | Simple | Cannot learn complex patterns | Regression output layer |
| Sigmoid | 1/(1+e^-x) | (0,1) | Yes | Smooth, probability output | Vanishing gradient | Binary classification |
| Tanh | tanh(x) | (-1,1) | Yes | Zero-centered | Vanishing gradient | Hidden layers |
| ReLU | max(0,x) | [0,∞) | Partially | Fast and simple | Dying ReLU problem | Deep neural networks |
| Leaky ReLU | x if x>0 else 0.01x | (-∞,∞) | Partially | Solves dying ReLU | Leak factor required | Hidden layers |
| ELU | x if x≥0 else e^x−1 | (-1,∞) | Yes | Smooth negative values | Computationally expensive | Deep learning |
| Softplus | ln(1+e^x) | (0,∞) | Yes | Smooth ReLU | Slower | Hidden layers |
| Swish | x/(1+e^-x) | (-0.3,∞ approx.) | Yes | Better gradient flow | Expensive | Modern deep learning |
| Softmax | e^xi/Σe^xj | (0,1) | Yes | Probability distribution | Only for output layer | Multi-class classification |

```python
import tensorflow as tf
from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten, LeakyReLU
import matplotlib.pyplot as plt
import time
```

```python
(x_train, y_train), (x_test, y_test) = mnist.load_data()

x_train = x_train / 255.0
x_test = x_test / 255.0

def create_model(activation):
    model = Sequential()

    model.add(Flatten(input_shape=(28,28)))

    if activation == "leaky_relu":
        model.add(Dense(128))
        model.add(LeakyReLU(negative_slope=0.01))
    else:
        model.add(Dense(128, activation=activation))

    model.add(Dense(10, activation='softmax'))

    model.compile(
        optimizer='adam',
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )

    return model

def train_model(activation):

    print("\nTraining using", activation)

    model = create_model(activation)

    start = time.time()

    history = model.fit(
        x_train,
        y_train,
        epochs=5,
        batch_size=128,
        validation_split=0.2,
        verbose=1
    )

    end = time.time()

    loss, accuracy = model.evaluate(x_test, y_test, verbose=0)

    print("Test Accuracy :", accuracy)
    print("Test Loss     :", loss)
    print("Training Time :", end - start)
    print("Validation Accuracy :", history.history['val_accuracy'][-1])

activations = ["sigmoid", "tanh", "relu", "leaky_relu"]

for act in activations:
    train_model(act)
```

# Task 7: Real-World Case Study

## Objective
Recommend the most suitable activation function for each real-world application and justify the choice.

| Application | Recommended Activation Function | Justification |
|-------------|---------------------------------|---------------|
| House Price Prediction | **Linear (Identity)** | House price prediction is a regression problem where the output is a continuous numerical value. The Linear activation function allows outputs in the range (-∞, ∞), making it ideal for predicting prices. |
| Cat vs Dog Classification | **ReLU (Hidden Layer)** + **Sigmoid (Output Layer)** | This is a binary classification problem. ReLU provides faster training and reduces the vanishing gradient problem, while Sigmoid produces a probability between 0 and 1 for the two classes. |
| Sentiment Analysis | **ReLU (Hidden Layer)** + **Sigmoid/Softmax (Output Layer)** | ReLU efficiently learns features from text data. Sigmoid is used for binary sentiment classification (positive/negative), while Softmax is used for multi-class sentiment classification. |
| Face Recognition | **ReLU** | Deep Convolutional Neural Networks commonly use ReLU because it accelerates training and effectively extracts facial features. |
| Object Detection | **Leaky ReLU** | Leaky ReLU prevents the dying ReLU problem by allowing a small gradient for negative inputs, improving object detection performance. |
| Time Series Forecasting | **Tanh** | Tanh outputs values between -1 and 1 and is widely used in RNNs and LSTMs to model sequential and time-dependent data. |
| Medical Image Classification | **ReLU** | Medical image classification requires deep CNNs. ReLU improves convergence speed and helps learn complex image features efficiently. |
| Fraud Detection | **ReLU (Hidden Layer)** + **Sigmoid (Output Layer)** | Fraud detection is generally a binary classification problem. ReLU learns complex patterns, while Sigmoid predicts the probability of fraud. |

## Conclusion

- **Linear** is best suited for regression problems.
- **Sigmoid** is preferred for binary classification output layers.
- **Softmax** is used for multi-class classification.
- **ReLU** is the most widely used activation function for hidden layers in deep neural networks.
- **Leaky ReLU** overcomes the dying ReLU problem.
- **Tanh** performs well in recurrent neural networks for sequential data such as time series forecasting.
